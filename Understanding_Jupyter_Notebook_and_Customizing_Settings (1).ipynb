## 1. What Is Jupyter Notebook?

**Jupyter Notebook** is an interactive, web-based environment for writing and running code. It is widely used in Python programming, data analysis, machine learning, scientific computing, and education.

A notebook combines executable code, the output of that code, visualizations, explanatory text, equations, and tables in one document.

### Why is it useful?
- **Interactive execution:** run one cell at a time.
- **Immediate feedback:** inspect values and errors as you learn.
- **Documentation:** explain the code with headings and text.
- **Visualization:** display charts and plots alongside code.
- **Experimentation:** test ideas without running an entire script from the beginning.

### Notebook file format
Jupyter notebooks are commonly saved with the **`.ipynb`** extension. The file stores cells, their source code, outputs, and metadata in a JSON-based format.

**Example:** `data_analysis.ipynb`

## 2. Jupyter Notebook vs. JupyterLab

- **Jupyter Notebook:** the classic notebook interface, focused on working with notebook documents.
- **JupyterLab:** a more flexible workspace that can display notebooks, terminals, text files, and other tools in tabs or panels.

The exact menus and settings can differ by Jupyter version and installation. A shortcut or menu item in the classic interface may not appear in precisely the same place in JupyterLab.

## 3. Main Parts of the Interface

1. **Notebook name:** the current notebook's filename.
2. **Menu bar:** menus for file operations, editing, viewing, running cells, and settings.
3. **Toolbar:** quick actions such as adding a cell, running a cell, interrupting execution, and restarting the kernel.
4. **Cells:** individual blocks containing code or text.
5. **Output area:** displays printed text, returned values, tables, errors, or charts.
6. **Kernel indicator:** shows the execution process associated with the notebook.
7. **File browser:** in JupyterLab, the file browser helps navigate files and folders.
8. **Status indicators:** may show whether the kernel is busy or idle and whether changes have been saved.

### A first code cell

````python
message = "Welcome to Jupyter Notebook!"
print(message)
````

## 4. Cell Types

### 4.1 Code cell
A Code cell contains executable code. Run it to execute the code using the active kernel.

### 4.2 Markdown cell
A Markdown cell contains formatted documentation. Markdown supports headings, bold and italic text, lists, links, tables, and many mathematical expressions.

Example Markdown source:
```markdown
# Data Analysis
This notebook explores **student performance**.

- Load the data
- Calculate summary statistics
- Visualize the results
```

### 4.3 Raw cell
A Raw cell contains text that is not processed as normal Markdown. It is useful in certain export or conversion workflows and is less common in everyday beginner work.

## 5. Essential Keyboard Shortcuts

Jupyter has two main interaction modes in the classic interface:

- **Command mode:** select and manage cells. Press `Esc` to enter it.
- **Edit mode:** type inside a cell. Press `Enter` to enter it.

Shortcuts can vary between the classic Notebook, JupyterLab, operating systems, and customized keymaps. Check the Help menu or keyboard-shortcut settings in your own installation.

| Shortcut (classic interface) | Action |
|---|---|
| `Shift + Enter` | Run the cell and select the next cell |
| `Ctrl + Enter` | Run the selected cell and stay on it |
| `Alt + Enter` | Run the cell and insert/select a cell below |
| `Esc` | Enter Command mode |
| `Enter` | Enter Edit mode |
| `A` in Command mode | Insert a cell above |
| `B` in Command mode | Insert a cell below |
| `M` in Command mode | Change the cell to Markdown |
| `Y` in Command mode | Change the cell to Code |
| `D`, then `D` in Command mode | Delete the selected cell |

**Teaching tip:** shortcuts such as `A`, `B`, `M`, `Y`, and `D,D` generally require Command mode. If you type them while editing a cell, they may appear as text instead.

## 6. Understanding the Kernel

A **kernel** is the process that executes code for a notebook. For a Python notebook, the kernel usually runs a Python interpreter and keeps variables in memory between cell executions.

### Example: variables persist in the kernel

````python
course_name = "Data Science"
students_count = 25
````

````python
print(course_name)
print(students_count)
````

The second cell can access the variables created by the first cell because both cells use the same running kernel.

### Kernel actions
- **Interrupt:** attempts to stop a cell that is currently running.
- **Restart:** starts a fresh kernel and clears variables from memory.
- **Restart and Run All:** restarts the kernel and executes the notebook cells from top to bottom (availability and wording vary by interface).

### Important execution-order issue
Jupyter lets you run cells out of order. That is convenient for exploration, but it can make results confusing. A notebook may depend on a variable created by a cell that appears later in the document.

**Best practice:** when checking that a notebook is reproducible, restart the kernel and run all cells from top to bottom.

## 9. Installing and Importing Packages

A Python **package** adds functionality beyond the standard library. For example, Pandas helps with tabular data, while Matplotlib helps create charts.

### Install a package from a notebook

````python
# Run this in a notebook cell when installation is needed:
%pip install pandas matplotlib
````

`%pip` is an IPython magic command designed to install packages into the Python environment associated with the active kernel. Internet access and permission to install packages may be required.

### Import packages

````python
import pandas as pd
import matplotlib.pyplot as plt
````

**Installation vs. import**
- `pip install pandas` installs a package into an environment.
- `import pandas as pd` makes the installed package available to the current Python session.

If an import fails after installation, check that the notebook kernel is using the environment where the package was installed. Restarting the kernel can sometimes help after environment changes.

## 10. Useful IPython Magic Commands

Magic commands are special commands provided by IPython. Line magics begin with `%`; cell magics begin with `%%`.

| Command | Purpose |
|---|---|
| `%pwd` | Display the current working directory |
| `%cd folder_path` | Change the working directory |
| `%whos` | List variables currently defined in the interactive session |
| `%time expression` | Measure the execution time of one expression |
| `%timeit expression` | Repeatedly measure an expression to estimate its execution time |
| `%matplotlib inline` | In classic IPython workflows, request plots to display inline |
| `%%time` | Measure the execution time of an entire cell |

### Examples

````python
%pwd
````

````python
values = [number * number for number in range(1000)]
%timeit sum(values)
````

````python
student = "Aarav"
marks = 88
%whos
````

Magic commands are specific to IPython-based environments; they are not ordinary Python syntax and may not work in a regular `.py` script.

## 11. Saving and Exporting a Notebook

### Save as a notebook
Save the document as a `.ipynb` file. This preserves its cell structure and may preserve outputs.

### Common export formats
- **`.py`** — Python script; formatting and Markdown may be converted to comments or otherwise transformed.
- **`.html`** — shareable web page, often preserving visible outputs.
- **`.pdf`** — availability depends on the installation and export tools; PDF export may require extra dependencies.
- **`.ipynb`** — the native editable notebook format.

Export options vary by version and environment. Look in the **File** menu for export or download options.

### Sharing checklist
- Run the notebook from top to bottom.
- Remove accidental outputs or sensitive information.
- Ensure required datasets and packages are documented.
- Use clear filenames, such as `01_python_basics.ipynb`.
- If sharing with others, explain any external files they need.

## 12. Practical Demonstration: Student Marks Analysis

In this activity, we will:
1. Create a small dataset.
2. Convert it into a Pandas DataFrame.
3. Calculate summary statistics.
4. Identify the highest score.
5. Create a bar chart.

### Step 1: Create the dataset

````python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["Aarav", "Diya", "Kabir", "Meera", "Rohan"],
    "Marks": [78, 92, 85, 88, 74]
}

df = pd.DataFrame(data)
df
````

### Step 2: Calculate summary statistics

````python
print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())
````

### Step 3: Find the top-performing student

````python
top_student = df.loc[df["Marks"].idxmax()]
print("Top student:", top_student["Student"])
print("Marks:", top_student["Marks"])
````

### Step 4: Visualize the marks

````python
plt.figure(figsize=(8, 4))
plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
````

- What does `pd.DataFrame(data)` do?
- Why do we use `df["Marks"].mean()`?
- What does `idxmax()` return?
- Why is a bar chart suitable for comparing individual students?
- What would happen if the `Marks` column contained missing values?

## 13. Common Problems and Solutions

| Problem | Possible solution |
|---|---|
| A cell appears to run forever | Check the code; interrupt the kernel if needed |
| `NameError` for a variable | Run the cell that defines it, or restart and run all cells in order |
| `ModuleNotFoundError` | Install the package in the active kernel's environment and check the environment |
| File not found | Check `os.getcwd()`, the file name, and the relative or absolute path |
| Changes are missing | Save the notebook and confirm the correct file is open |
| Output looks outdated | Re-run the relevant cell and inspect execution order |
| Notebook is slow | Reduce unnecessary output, avoid very large displays, and profile expensive code |

## 14. Frequently Asked Student Questions

**Q1. Is Jupyter Notebook a programming language?**  
No. It is an environment for running code written in languages such as Python.

**Q2. What is the difference between a notebook and a Python file?**  
A notebook organizes work into cells and can store rich outputs. A `.py` file is a plain Python script, usually executed as a program.

**Q3. What is a kernel?**  
The process that executes notebook code and maintains session state, including variables in memory.

**Q4. Does restarting the kernel delete the notebook?**  
No. It clears the running session's memory. Saved code remains in the notebook, but variables must be recreated by rerunning cells.

**Q5. Why can cells produce unexpected results when run out of order?**  
A cell may depend on variables or state created by previously executed cells, even if those cells appear elsewhere in the document.

**Q6. Can Jupyter be used for machine learning?**  
Yes. It is commonly used for data exploration, preprocessing, experiments, model evaluation, and visualization.

**Q7. Should every task go into one cell?**  
No. Small, logically grouped cells make a notebook easier to understand, test, and debug.

**Q8. What is the difference between a Code cell and a Markdown cell?**  
A Code cell executes code; a Markdown cell displays formatted explanatory text.
